```python
import matplotlib.pyplot as plt
import numpy as np

def cost_func(x):
    return (x+3)**2


def gradient(x):
    return 2*(x+3)
```

```python
x_values = []
y_values = []
num_iterations = 100

x_initial = 2
x = x_initial
lr = 0.1

for i in range(num_iterations):
    x_values.append(x)
    y_values.append(cost_func(x))
    
    x = x - lr * gradient(x)
    print(f"Iteration {i+1}: x_value = {x}, cost = {cost_func(x)}")

print(f'Optimal x : {x}')
```

```python
x_coordinates = np.linspace(-5, 5, 100)
plt.plot(x_coordinates, cost_func(x_coordinates), label='Cost Function')
plt.xlabel('x')
plt.ylabel('Cost')
plt.title("Function graph for the cost function : (x+3)^2")
plt.plot(2, cost_func(2), 'ro')
plt.show()
```

```python
plt.plot(x_coordinates, cost_func(x_coordinates), label='Cost Function')
plt.plot(x_values, y_values, 'ro-')
plt.xlabel('x')
plt.ylabel('Cost')
plt.title("Gradient descent for the cost function : (x+3)^2")
plt.show()
```
